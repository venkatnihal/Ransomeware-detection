```python
import time
import numpy as np
from sklearn.ensemble import RandomForestClassifier
from sklearn.model_selection import train_test_split

# Medium dataset
medium_data = data.sample(500000, random_state=42)

X_med = medium_data.drop('Label',axis=1)

X_med = X_med.select_dtypes(include=['int64','float64'])

y_med = medium_data['Label']

# Split
X_train_m,X_test_m,y_train_m,y_test_m = train_test_split(
X_med,y_med,test_size=0.2,random_state=42)

# Clean dataset
X_train_m = np.nan_to_num(X_train_m,nan=0,posinf=0,neginf=0)
X_test_m = np.nan_to_num(X_test_m,nan=0,posinf=0,neginf=0)

# Limit large values
X_train_m[X_train_m>1e9]=1e9
X_test_m[X_test_m>1e9]=1e9

X_train_m[X_train_m<-1e9]=-1e9
X_test_m[X_test_m<-1e9]=-1e9

# Train model
start=time.time()

rf_medium = RandomForestClassifier()

rf_medium.fit(X_train_m,y_train_m)

end=time.time()
    
medium_time=end-start

print("Medium Dataset Time:",medium_time)
```
